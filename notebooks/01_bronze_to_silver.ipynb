## 2. Null Handling Strategy

Following strict Medallion Architecture principles, the Silver layer must serve **all downstream consumers**, not just the credit risk model. Therefore, we avoid taking business or modeling decisions that irreversibly destroy data.

**1. Semantic Zero-Imputation (Counts)**
For pure count/event flags (e.g., `num_tl_120dpd_2m`, `pub_rec`, `acc_now_delinq`), a missing value universally means the bureau found zero occurrences. We safely impute these with `0`.

**2. Strict Preservation of Magnitudes and Ratios**
For financial magnitudes (e.g., `tot_cur_bal`, `bc_open_to_buy`) and ratios (`dti`, `revol_util`), missing values are kept as native `NaN`. We do **NOT** drop these rows in Silver, as doing so would destroy perfectly good history for consumers who don't use those specific columns. Row-drops are deferred to the Gold layer.

**3. Preservation of Age Variables**
For "months since" variables (`mo_sin_*` and `mths_since_*`), we keep them as native numeric `NaN` (preserving `float64`). We do not force arbitrary bins or sentinels like `-999`, leaving the choice of handling missingness to the downstream ML algorithms in Gold.

### Column Selection
Drop columns with >= 20% nulls (see markdown above for why).

```python

high_null_cols = null_pct[null_pct >= 20].index.tolist()

df_silver = df.drop(columns=high_null_cols, errors='ignore')
print(f"Shape after high-null column drop: {df_silver.shape}")


# Capture a copy BEFORE imputation for the validation check at the end of the notebook
df_silver_raw_for_test = df_silver.copy()
```

### Technical Drops
Drop universal garbage and redundancies. `zip_code` is kept for Gold layer segmentation.

```python
silver_drops = [
    'url', 'policy_code', 'title', 'emp_title', 'desc',
    'funded_amnt_inv', 'out_prncp_inv', 'total_pymnt_inv',
]
df_silver = df_silver.drop(columns=silver_drops, errors='ignore')
print(f"Shape after technical drops: {df_silver.shape}")

```

```python
# ── Technical Imputation ───────────────────────────────────────────


# B) Employment length: empty = unemployed / not reported
if 'emp_length' in df_silver.columns:
    df_silver['emp_length'] = df_silver['emp_length'].fillna('Not Disclosed')

```

```python
# ── Memory Optimization ────────────────────────────────────────────
categorical_cols = [
    'term', 'grade', 'sub_grade', 'emp_length', 'home_ownership',
    'verification_status', 'loan_status', 'purpose', 'application_type',
]
cols_to_cast = [col for col in categorical_cols if col in df_silver.columns]
for col in cols_to_cast:
    if col == 'emp_length':
        emp_cats = ['< 1 year', '1 year', '2 years', '3 years', '4 years', '5 years', '6 years', '7 years', '8 years', '9 years', '10+ years', 'Not Disclosed']
        df_silver[col] = pd.Categorical(df_silver[col], categories=emp_cats, ordered=False)
    else:
        df_silver[col] = df_silver[col].astype('category')

```

### Final Null Check
Checking remaining nulls before applying the targeted final imputation.


### Final Sentinel-Safe Imputations

To finalize the Silver layer, we handle the remaining missing values based strictly on their semantic meaning, ensuring no business decisions are irreversibly baked into this layer:

1. **Count Fields (`num_*`)**: Imputed with `0`. In credit risk reporting, a missing count for derogatory events or specific account types universally means the bureau found zero occurrences to report.
2. **Age & Recency Fields (`mo_sin_*`, `mths_since_*`)**: Left as **Null**. Imputing these with sentinels like `-999` would silently corrupt downstream averages, while binning them into categories prematurely locks in a modeling choice. They remain native `float64` `NaN`s.
3. **Magnitudes (`tot_*`, `avg_*`, `pct_*`)**: Left as **Null**. Imputing these would fabricate financial data and irreversibly bias downstream calculations. The decision to drop rows based on missing magnitudes is a use-case specific decision deferred to the Gold layer.

```python
null_pct = (df_silver.isnull().sum() / len(df_silver) * 100).sort_values(ascending=False)
print(null_pct[null_pct > 0].round(2))
```

### Intentional Nulls Deferred to Gold

The following columns still contain nulls. As per strict Medallion Architecture principles, these are all **magnitudes, ratios, or unbinned age variables**.

We intentionally **DO NOT** drop these rows nor impute them in Silver because doing so would irreversibly destroy data for other downstream consumers. The decision to drop rows containing these nulls, or to bin them, is a use-case specific modeling decision that belongs exclusively in the **Gold layer**.

**Specific Treatment Notes:**
- `last_pymnt_d`, `last_credit_pull_d`, `earliest_cr_line`: Nulls reflect recency or absence of history (e.g., no payment cycle yet), treated as right-censoring in Gold, not a data quality gap.
- `annual_inc`: Although nulls are near trace levels (0.00%), it is a critical magnitude for risk modeling and thus preserved as `NaN` rather than imputed.
- `total_acc`, `open_acc`: While technically counts, a true zero here is a logical anomaly (virtually everyone has at least one account). They are deferred as ambiguous missing data.
- `zip_code`: Deferred as a conscious business segmentation decision.

```python
# Eventos derogatorios: seguro imputar con 0
safe_zero_impute = [
    'delinq_amnt', 'acc_now_delinq', 'inq_last_6mths', 'delinq_2yrs', 'pub_rec',
    'pub_rec_bankruptcies', 'tax_liens', 'chargeoff_within_12_mths', 'collections_12_mths_ex_med'
]

# Conteos de tradelines/buro: el nulo es 'no tenemos historial extendido de esta cohorte'
cohort_deferred_counts = [
    'num_op_rev_tl', 'num_tl_90g_dpd_24m', 'num_rev_accts', 'num_tl_120dpd_2m',
    'num_actv_rev_tl', 'num_accts_ever_120_pd', 'num_bc_tl', 'num_bc_sats',
    'num_tl_30dpd', 'num_actv_bc_tl', 'num_rev_tl_bal_gt_0', 'num_sats',
    'num_il_tl', 'num_tl_op_past_12m', 'mort_acc', 'acc_open_past_24mths'
]

for col in safe_zero_impute:
    if col in df_silver.columns:
        df_silver[col] = df_silver[col].fillna(0)
        
# cohort_deferred_counts se deja intacto (NaN nativo) para tratarlo en Gold
```

```python
final_nulls = (df_silver.isnull().sum() / len(df_silver) * 100).sort_values(ascending=False)
final_nulls = final_nulls[final_nulls > 0]
print("Final intentional nulls sent to Gold (%):")
print(final_nulls.round(2))

# Strict grouped validation of permitted nulls
expected_null_cols = {
    'counts_bureau_cohort': [
        'num_op_rev_tl', 'num_tl_90g_dpd_24m', 'num_rev_accts', 'num_tl_120dpd_2m',
        'num_actv_rev_tl', 'num_accts_ever_120_pd', 'num_bc_tl', 'num_bc_sats',
        'num_tl_30dpd', 'num_actv_bc_tl', 'num_rev_tl_bal_gt_0', 'num_sats',
        'num_il_tl', 'num_tl_op_past_12m', 'mort_acc', 'acc_open_past_24mths'
    ],
    'ratios_pct': ['revol_util', 'bc_util', 'percent_bc_gt_75', 'pct_tl_nvr_dlq'],
    'magnitudes_usd': [
        'tot_cur_bal', 'avg_cur_bal', 'bc_open_to_buy', 'tot_coll_amt', 
        'total_rev_hi_lim', 'tot_hi_cred_lim', 'total_il_high_credit_limit', 
        'total_bal_ex_mort', 'total_bc_limit', 'annual_inc'
    ],
    'ratios_other': ['dti'],
    'months_since': [
        'mths_since_recent_inq', 'mths_since_recent_bc',
        'mo_sin_old_il_acct', 'mo_sin_rcnt_rev_tl_op',
        'mo_sin_old_rev_tl_op', 'mo_sin_rcnt_tl'
    ],
    'recency_date': ['last_pymnt_d', 'last_credit_pull_d', 'earliest_cr_line'],
    'business_deferred': ['zip_code'],
    'counts_ambiguous_zero': ['total_acc', 'open_acc']
}
all_expected = set(sum(expected_null_cols.values(), []))
actual_null_cols = set(df_silver.columns[df_silver.isnull().any()])

assert actual_null_cols == all_expected, f"Mismatch!\nUnexpected nulls: {actual_null_cols - all_expected}\nMissing expected nulls: {all_expected - actual_null_cols}"
print("\nAssert passed: All remaining nulls are explicitly authorized.")
```

## 3. Validation — why NOT to dropna() here

One check done to justify the decision above (kept for documentation / reproducibility,
not part of the main pipeline — safe to skip on re-run).



Check A: does a dropna() here introduce temporal bias?
(i.e. does it disproportionately wipe out early loan-issue years?)


```python
df['issue_d_year'] = pd.to_datetime(df['issue_d'], format='%b-%Y').dt.year
before_dist = df['issue_d_year'].value_counts(normalize=True).sort_index()

df_silver_dropna_test = df_silver_raw_for_test.dropna().copy()
df_silver_dropna_test['issue_d_year'] = pd.to_datetime(
    df_silver_dropna_test['issue_d'], format='%b-%Y'
).dt.year
after_dist = df_silver_dropna_test['issue_d_year'].value_counts(normalize=True).sort_index()

comparison = pd.DataFrame({'antes': before_dist, 'despues': after_dist})
comparison['diff_pp'] = (comparison['despues'] - comparison['antes']) * 100
print("Temporal distribution shift if dropna() were applied:")
print(comparison.round(3))
```

Result: 2007-2011 rows disappear almost entirely -> confirms MNAR structural pattern,
not random noise. Dropping rows here would erase the early historical record.
